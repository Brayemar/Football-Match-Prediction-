```python
import pandas
data = pandas.read_csv("datasets/matchprediction_data.csv")
data
```

```python
from sklearn.preprocessing import LabelEncoder

home_le = LabelEncoder()
away_le = LabelEncoder()
result_le = LabelEncoder()

data['home_team'] = home_le.fit_transform(data['home_team'])
data['away_team'] = home_le.fit_transform(data['away_team'])
data['result'] = home_le.fit_transform(data['result'])

```

```python
data.describe()
```

```python
X = data.drop('result', axis = 1)
Y = data['result']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier()
model.fit(X_train, Y_train)
```

```python
# --- TEAM MAP ---
teams = sorted(data['home_team'].unique())
team_map = {i: team for i, team in enumerate(teams)}

print("\n⚽ MATCH PREDICTION SYSTEM ⚽")

# --- SHOW MENU WITH NUMBER + NAME ---
print("\nSelect a team by number:\n")

for i, team in team_map.items():
    print(f"[{i}] {team}")

# --- USER CHOICE ---
home_team = int(input("\nChoose HOME team number: "))
away_team = int(input("Choose AWAY team number: "))

# --- CONFIRM SELECTION ---
print("\n✔ You selected:")
print(f"Home: [{home_team}] {team_map[home_team]}")
print(f"Away: [{away_team}] {team_map[away_team]}")


home_possession = float(input("Home possession: "))
away_possession = float(input("Away possession: "))
home_xg = float(input("Home expected goals: "))
away_xg = float(input("Away expected goals: "))
home_shots = float(input("Home shots per 90: "))
away_shots = float(input("Away shots per 90: "))
home_pass = float(input("Home pass accuracy: "))
away_pass = float(input("Away pass accuracy: "))
home_form = float(input("Home form: "))
away_form = float(input("Away form: "))
h2h_home = float(input("H2H home win rate: "))
h2h_draw = float(input("H2H draw rate: "))
h2h_away = float(input("H2H away win rate: "))
home_injuries = float(input("Home injuries: "))
away_injuries = float(input("Away injuries: "))
home_adv = float(input("Home advantage: "))
goal_diff = float(input("Goal difference average: "))

# --- BUILD SAMPLE ---
sample = [[
    home_team,
    away_team,
    home_possession,
    away_possession,
    home_xg,
    away_xg,
    home_shots,
    away_shots,
    home_pass,
    away_pass,
    home_form,
    away_form,
    h2h_home,
    h2h_draw,
    h2h_away,
    home_injuries,
    away_injuries,
    home_adv,
    goal_diff
]]

# --- PREDICTION ---
prediction = model.predict(sample)[0]
result_name = result_le.inverse_transform([prediction])[0]

# --- OUTPUT (converted back to names) ---
print("\n🏁 MATCH RESULT PREDICTION 🏁")
print("Home Team:", team_map[home_team])
print("Away Team:", team_map[away_team])
print("Predicted Result:", result_name)
```

```python
predictions = model.predict(X_test)
print(predictions)
print("We expected", Y_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(Y_test, predictions)
print("Accuracy:", accuracy)
```
