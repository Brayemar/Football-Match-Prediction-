```python
import pandas
data = pandas.read_csv("datasets/matchprediction_data.csv")
data
```

```python
# Here I described the data
data.describe()
```

```python
# Here I now converted the results from text to number form
# Whereby the home_win = 2, draw = 1 and away_win = 0

data["result"] = data["result"].map({
    "home_win": 2,
    "draw" : 1,
    "away_win" : 0
})
```

```python
# Here I now donverted my datasets into an array format
array = data.values
# print(array)

# Here I tell the system what features it is to take where by I tell it to pic all rows(:) from column 2(2)(but not including) up to the last column(-1)
# This is like slicing
X = array[:, 2:-1]
# X

# Here I tell the system what I want it to try and predict and in this case it is the results(home_win = 2, draw - 1, away_win = 0)
Y = array[:, -1]
Y
```

```python
# At this point I went ahead and split the data into training(X_train,Y_train) and testing(X_test,Y_test) sets.

from sklearn import model_selection

X_train, X_test, Y_train,Y_test = model_selection.train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
# Now I trained the model
# First of all I began by importing the model
from sklearn.tree import DecisionTreeClassifier

# Create the model
model = DecisionTreeClassifier()

# Finally I actually train the model
model.fit(X_train, Y_train)
```
