```python
# Import required libraries
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

```

```python
# Here I load the dataset for training the model
data = pd.read_csv("datasets/matchprediction_data.csv")
print(data)
```

```python
# I ENCODE THE DATA CATEGORICALLY
# Convert team names and results into numerical values

team_le = LabelEncoder()
result_le = LabelEncoder()

# Combine home and away teams to ensure consistent encoding
all_teams = pd.concat([data['home_team'], data['away_team']])
team_le.fit(all_teams)


# Transform using the SAME fitted encoder
data['home_team'] = team_le.transform(data['home_team'])
data['away_team'] = team_le.transform(data['away_team'])

# Encode result
data['result'] = result_le.fit_transform(data['result'])

#print(team_le.classes_)
```

```python
data.describe()
```

```python
# DEFINE FEATURES AND TARGET

# X = inputs (teams), y = output (result)
X = data[[
    'home_team',
    'away_team',
    'home_possession',
    'away_possession',
    'home_expected goals',
    'away_expected goals',
    'home_shots_per_ninety',
    'away_shots_per_ninety',
    'home_pass_accuracy',
    'away_pass_accuracy',
    'home_form',
    'away_form',
    'h2h_home_win_rate',
    'h2h_draw_rate',
    'h2h_away_win_rate',
    'home_injuries',
    'away_injuries',
    'home_advantage',
    'goal_difference_average'
]]
Y = data['result']

# print(X)
# print(Y)
```

```python
# SPLIT DATA
# Split dataset into training and testing sets

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
# ===============================
# MODEL TRAINING
# ===============================
# We use Decision Tree Classifier to predict match outcomes
# The dataset is split into training and testing sets

model = DecisionTreeClassifier()
model.fit(X_train, Y_train)
```

```python
# Test model to see what it would predict
predictions = model.predict(X_test)

print("Model Predicted:", predictions)
print("We expected:", list(Y_test))
```

```python
# Now we calculate the accuracy of the model
accuracy = accuracy_score(Y_test, predictions)
print("Accuracy of the Model is: ", accuracy)
```

```python

# ===============================
# ⚽ TEAM MENU (ID + NAME)
# ===============================
teams = list(team_le.classes_)

print("⚽ MATCH PREDICTION SYSTEM ⚽")
print("Select a team using the ID below:")

for i, team in enumerate(teams):
    print(f"[{i}] {team}")

# ===============================
# USER INPUT
# ===============================
home_id = int(input("Enter HOME team ID: "))
away_id = int(input("Enter AWAY team ID: "))

# Validate input
if home_id not in range(len(teams)) or away_id not in range(len(teams)):
    print("Invalid team ID selected. Please restart.")
    exit()

# ===============================
# CONVERT ID → TEAM NAME
# ===============================
home_team_name = teams[home_id]
away_team_name = teams[away_id]

# ===============================
# ENCODE FOR MODEL
# ===============================
home_encoded = team_le.transform([home_team_name])[0]
away_encoded = team_le.transform([away_team_name])[0]

# ===============================
# CONFIRM SELECTION
# ===============================
print("YOU SELECTED:")
print(f"Home: [{home_id}] {home_team_name}")
print(f"Away: [{away_id}] {away_team_name}")

# ===============================

# ===============================
# AUTO-FILL MATCH STATS
# ===============================

home_matches = data[data['home_team'] == home_encoded]
away_matches = data[data['away_team'] == away_encoded]

if len(home_matches) == 0 or len(away_matches) == 0:
    print("Not enough data for selected teams.")
    exit()

home_data = home_matches.mean(numeric_only=True)
away_data = away_matches.mean(numeric_only=True)

# ===============================
# BUILD SAMPLE (MODEL INPUT)
# ===============================
sample = [[
    home_encoded,
    away_encoded,
    home_data['home_possession'],
    away_data['away_possession'],
    home_data['home_expected goals'],
    away_data['away_expected goals'],
    home_data['home_shots_per_ninety'],
    away_data['away_shots_per_ninety'],
    home_data['home_pass_accuracy'],
    away_data['away_pass_accuracy'],
    home_data['home_form'],
    away_data['away_form'],
    home_data['h2h_home_win_rate'],
    home_data['h2h_draw_rate'],
    home_data['h2h_away_win_rate'],
    home_data['home_injuries'],
    away_data['away_injuries'],
    home_data['home_advantage'],
    home_data['goal_difference_average']
]]


# ===============================
# PREDICTION
# ===============================
prediction = model.predict(sample)[0]
result_name = result_le.inverse_transform([prediction])[0]

# ===============================
# FINAL OUTPUT
# ===============================
print("🏆 MATCH RESULT PREDICTION 🏆")

print(f"Home Team: [{home_id}] {home_team_name}")
print(f"Away Team: [{away_id}] {away_team_name}")

print(f"🏆🏆Predicted Result: {result_name}")
```
