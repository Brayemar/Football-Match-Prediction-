```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# We load the dataset to explore its structure and contents
data = pd.read_csv("datasets/matchprediction_data.csv")
data
```

```python

# Display first few rows to understand what the data looks like
print("First 5 rows of the dataset:")
print(data.head())
```

```python
# Show structure: column names, data types, non-null counts
print("Dataset Information:")
print(data.info())
```

```python
# Check for missing values in each column
print("Missing Values:")
print(data.isnull().sum())
```

```python
# GRAPH 1: MATCH RESULT DISTRIBUTION
# Shows how many home wins, away wins, and draws exist
data['result'].value_counts().plot(kind='bar')
plt.title("Match Result Distribution")
plt.xlabel("Result Type")
plt.ylabel("Number of Matches")
plt.show()
```

```python
# GRAPH 2: RESULT PERCENTAGE (PIE CHART)
# Helps visualize proportions of outcomes
data['result'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Match Result Share (%)")
plt.ylabel("")  # Remove default ylabel
plt.show()
```

```python
# GRAPH 3: TOP 10 HOME TEAMS
# Shows which teams appear most often as home teams
data['home_team'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Home Teams by Matches Played")
plt.xlabel("Team")
plt.ylabel("Number of Matches")
plt.xticks(rotation=45)
plt.show()
```

```python
# GRAPH 4: TOP 10 AWAY TEAMS
# Shows which teams appear most often as away teams
data['away_team'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Away Teams by Matches Played")
plt.xlabel("Team")
plt.ylabel("Number of Matches")
plt.xticks(rotation=45)
plt.show()
```

```python
# GRAPH 5: HOME VS AWAY VS DRAW COMPARISON
# Direct comparison of match outcomes
result_counts = data['result'].value_counts()

plt.bar(result_counts.index, result_counts.values)
plt.title("Home Wins vs Away Wins vs Draws")
plt.xlabel("Result Type")
plt.ylabel("Count")
plt.show()
```

```python
# GRAPH 6: HOME VS AWAY MATCH FREQUENCY
# Compares how often teams play home vs away matches
home_counts = data['home_team'].value_counts()
away_counts = data['away_team'].value_counts()

plt.scatter(home_counts, away_counts)
plt.title("Home vs Away Match Frequency")
plt.xlabel("Home Matches")
plt.ylabel("Away Matches")
plt.show()
```

```python
# GRAPH 7: PERFORMANCE OF TOP TEAM (HOME)
# Analyze results of the most frequent home team
top_team = data['home_team'].value_counts().idxmax()
team_data = data[data['home_team'] == top_team]

team_data['result'].value_counts().plot(kind='bar')
plt.title(f"Results of {top_team} in Home Matches")
plt.xlabel("Result")
plt.ylabel("Count")
plt.show()
```

```python
# GRAPH 8: RESULT DISTRIBUTION (HISTOGRAM)
# Shows distribution pattern of match outcomes
data['result'].value_counts().plot(kind='hist')
plt.title("Histogram of Match Results")
plt.xlabel("Result Categories")
plt.ylabel("Frequency")
plt.show()
```
