```python
# IMPORT LIBRARIES
# ===============================
# pandas → used for handling datasets (tables)
# matplotlib → used for graphs and visualizations
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# ===============================
# LOAD DATASET
# ===============================
# We load the football match dataset into a DataFrame
# This allows us to analyze match statistics
data = pd.read_csv("datasets/matchprediction_data.csv")
data
```

```python
# ===============================
# BASIC DATA INSPECTION
# ===============================
# Show first 5 rows to understand structure of dataset
print("First 5 rows of dataset:")
print(data.head())
```

```python
# Show dataset structure (columns, data types, missing values)
print("Dataset Information:")
print(data.info())
```

```python
# Check for missing values (important for ML models)
print("Missing Values in Each Column:")
print(data.isnull().sum())
```

```python
# ===============================
# BASIC STATISTICS
# ===============================
# Gives mean, min, max, and distribution of numeric values
print("Statistical Summary:")
print(data.describe(include='all'))
```

```python
# ===============================
# MATCH RESULT DISTRIBUTION
# ===============================
# Shows how many home wins, away wins, and draws exist
# This helps detect dataset imbalance
data['result'].value_counts().plot(kind='bar')
plt.title("Match Result Distribution")
plt.xlabel("Result Type")
plt.ylabel("Number of Matches")
plt.show()
```

```python

# ===============================
# RESULT PERCENTAGE (PIE CHART)
# ===============================
# Shows percentage breakdown of match outcomes
# Helps visualize dominance of certain results
data['result'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Match Result Percentage")
plt.ylabel("")
plt.show()
```

```python
# ===============================
# TOP HOME TEAMS
# ===============================
# Shows which teams appear most often as home teams
# Helps understand dataset structure and bias
data['home_team'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Home Teams")
plt.xlabel("Team")
plt.ylabel("Matches Played")
plt.xticks(rotation=45)
plt.show()
```

```python

# ===============================
# TOP AWAY TEAMS
# ===============================
# Shows which teams appear most often as away teams
data['away_team'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Away Teams")
plt.xlabel("Team")
plt.ylabel("Matches Played")
plt.xticks(rotation=45)
plt.show()
```

```python
# ===============================
# POSSESSION vs RESULT
# ===============================
# Shows how ball possession affects match outcomes
# Helps understand if possession influences winning
data.groupby('result')[['home_possession', 'away_possession']].mean().plot(kind='bar')
plt.title("Average Possession by Result")
plt.ylabel("Possession %")
plt.show()
```

```python
# ===============================
# EXPECTED GOALS (xG) vs RESULT
# ===============================
# Shows how attacking quality (xG) affects outcomes
data.groupby('result')[['home_expected goals', 'away_expected goals']].mean().plot(kind='bar')
plt.title("Expected Goals vs Match Result")
plt.ylabel("xG")
plt.show()
```

```python

# ===============================
# SHOTS vs RESULT
# ===============================
# Shows if more shots lead to winning matches
data.groupby('result')[['home_shots_per_ninety', 'away_shots_per_ninety']].mean().plot(kind='bar')
plt.title("Shots per 90 vs Result")
plt.ylabel("Shots")
plt.show()
```

```python
# ===============================
# PASS ACCURACY vs RESULT
# ===============================
# Shows if passing accuracy influences match outcomes
data.groupby('result')[['home_pass_accuracy', 'away_pass_accuracy']].mean().plot(kind='bar')
plt.title("Pass Accuracy vs Result")
plt.ylabel("Accuracy %")
plt.show()
```

```python
# ===============================
# HOME POSSESSION DISTRIBUTION
# ===============================
# Shows how possession is distributed across matches
# This helps detect consistency of data
data['home_possession'].plot(kind='hist', bins=10)
plt.title("Distribution of Home Possession")
plt.xlabel("Possession %")
plt.ylabel("Frequency")
plt.show()
```
